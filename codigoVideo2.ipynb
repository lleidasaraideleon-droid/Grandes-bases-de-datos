```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import os
```

```python
os.chdir("/Users/lleid/Downloads")
         #"C:\Users\lleid\Downloads\Wholesale customers data.csv"
data = pd.read_csv('Wholesale customers data.csv')
data.head()
```

```python
from sklearn.preprocessing import normalize
data_scaled = normalize(data)
data_scaled
```

```python
data_scaled = pd.DataFrame(data_scaled, columns = data.columns)
data_scaled.head()
```

```python
import scipy.cluster.hierarchy as shc
plt.figure(figsize = (10,7))
plt.title("Dendograma")
dend = shc.dendrogram(shc.linkage(data_scaled, method = 'ward'))
```

```python
#python determina un balance en la distancia intra-cluster y la inter-cluster
print(dend['color_list'])
```

```python
colores_unicos = set(dend['color_list'])  #set toma elementos repetidos y los da en una lista 
colores_unicos
```

```python
num_clusters_optimo = len(colores_unicos) - 1   
num_clusters_optimo
```

```python
plt.figure(figsize = (10,7))
plt.title("Dendograma")
dend = shc.dendrogram(shc.linkage(data_scaled, method = 'ward'))
plt.axhline(y=6, color = 'r', linestyle = '--')
```

```python
data
```

```python
from sklearn.cluster import AgglomerativeClustering
cluster = AgglomerativeClustering(n_clusters=2, metric='euclidean', linkage='ward')
grupos = cluster.fit_predict(data_scaled)
grupos
```

```python
plt.figure(figsize = (10,7))
plt.scatter(data_scaled['Milk'],data_scaled['Grocery'], c = cluster.labels_)
```

```python
# Separación de grupos mediante PCA
campos = data_scaled.values
print(campos)
```

```python
from sklearn import decomposition
pca = decomposition.PCA(n_components = 2)
pca.fit(campos)
campos = pca.transform(campos)
campos
```

```python
plt.figure(figsize=(10,7))
plt.scatter(campos[:,0], campos[:,1], c = cluster.labels_)
```

```python
dataframe = pd.DataFrame(grupos,columns=['grupo'])
dataframe
```

```python
dataframe2 = pd.concat([data, dataframe], axis=1, join = 'inner' )
dataframe2
```

# Análisis de caso de Estudiantes

```python
df = pd.DataFrame(np.array([[7,6.5,9.2,8.6,8],
                           [7.5,9.4,7.3,7,7],
                           [7.6,9.2,8,8,7.5],
                           [5,6.5,6.5,7,9],
                           [6,6,7.8,8.9,7.3],
                           [7.8,9.6,7.7,8,6.5],
                           [6.3,6.4,8.2,9,7.2],
                           [7.9,9.7,7.5,8,6],
                           [6,6,6.5,5.5,8.7],
                           [6.8,7.2,8.7,9,7]]),
                 columns = ['Matematicas','Ciencias','Español','Historia','EdFisica'])
df
```

```python
from sklearn.preprocessing import normalize
data_scaled = normalize(df)
data_scaled
```

```python
import scipy.cluster.hierarchy as shc
plt.figure(figsize = (10,7))
plt.title("Dendograma")
dend = shc.dendrogram(shc.linkage(data_scaled, method = 'ward'))
```

```python
colores_unicos = set(dend['color_list'])
num_clusters_optimo = len(colores_unicos) - 1   
num_clusters_optimo
```

```python
plt.figure(figsize = (10,7))
plt.title("Dendograma")
dend = shc.dendrogram(shc.linkage(data_scaled, method = 'ward'))
plt.axhline(y=0.2, color = 'r', linestyle = '--')
```

```python
campos = data_scaled
from sklearn import decomposition
pca = decomposition.PCA(n_components = 2)
pca.fit(campos)
campos = pca.transform(campos)
campos
```

```python
from sklearn.cluster import AgglomerativeClustering
cluster = AgglomerativeClustering(n_clusters=3, metric='euclidean', linkage='ward')
grupos = cluster.fit_predict(data_scaled)
grupos
```

```python
plt.figure(figsize=(10,7))
plt.scatter(campos[:,0], campos[:,1], c = cluster.labels_)
```

```python
dataframe = pd.DataFrame(grupos,columns=['grupo'])
dataframe
```

```python
dataframe2 = pd.concat([df, dataframe], axis=1, join = 'inner' )
dataframe2
```
