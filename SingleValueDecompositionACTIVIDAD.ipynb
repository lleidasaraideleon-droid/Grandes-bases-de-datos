```python
import numpy as np
import pandas as pd
from scipy.linalg import svd
from numpy import diag, zeros
import matplotlib.pyplot as plt
from mpl_toolkits.mplot3d import Axes3D

# Base de datos de estudiantes 
data = {
    "Matematicas": [7.0, 7.5, 7.6, 5.0, 6.0, 7.8, 6.3, 7.9, 6.0, 6.8],
    "Ciencias":    [6.5, 9.4, 9.2, 6.5, 6.0, 9.6, 6.4, 9.7, 6.0, 7.2],
    "Espanol":     [9.2, 7.3, 8.0, 6.5, 7.8, 7.7, 8.2, 7.5, 6.5, 8.7],
    "Historia":    [8.6, 7.0, 8.0, 7.0, 8.9, 8.0, 9.0, 8.0, 5.5, 9.0],
    "EdFisica":    [8.0, 7.0, 7.5, 9.0, 7.3, 6.5, 7.2, 6.0, 8.7, 7.0],
}
estudiantes = ["Lucia","Pedro","Ines","Luis","Andres","Ana","Carlos","Jose","Sonia","Maria"]
A = pd.DataFrame(data, index=estudiantes)
A
```

```python
from numpy import diag
from numpy import zeros
from scipy.linalg import svd
import matplotlib.pyplot as plt

U, s, VT = svd(A) #obtiene las tres matrices cuyo producto arroja la matriz A
Sigma = zeros((A.shape[0], A.shape[1])) #matriz de ceros de la misma dimensión que A
Sigma[:A.shape[1], :A.shape[1]] = diag(s)
print("Matriz U:")
print(U)
print("Matriz Sigma")
print(Sigma)
print("Matriz VT")
print(VT)
# Para reconstruir la matriz original podemos probarlo
B = U.dot(Sigma.dot(VT))
print(B)
```

```python
#Reducción a dos dimensiones
n_elements = 2
UReduced2 = U[:, :n_elements]
print("U Reducida:")
print(UReduced2)
SigmaReduced2 = Sigma[:n_elements, :n_elements]
print("Matriz Sigma Reducida")
print(SigmaReduced2)
VTReduced2 = VT[:n_elements, :]
print("VT Reducida")
print(VTReduced2)
AReduced2 = UReduced2.dot(SigmaReduced2.dot(VTReduced2))
print("Matriz A transformada:")
print(AReduced2)
T2 = UReduced2.dot(SigmaReduced2)
print("Matriz T (2 Dimensiones):")
print(T2)
```

```python
#realizamos el gráfico ya con las etiquetas
df = pd.DataFrame(T2, columns=["a", "b"])
x = df.iloc[:,0]
y = df.iloc[:, 1]
x = x.to_numpy()
y = y.to_numpy()
fig, ax = plt.subplots()
ax.set_title("Mapa de observaciones (SVD)")
ax.scatter(x,y)
for i, txt in enumerate(estudiantes):
    ax.annotate(txt, (x[i], y[i]))
```

```python
#Reducción a tres dimensiones
n_elements = 3

UReduced3 = U[:, :n_elements]

SigmaReduced3 = Sigma[:n_elements, :n_elements]

VTReduced3 = VT[:n_elements, :]

AReduced3 = UReduced3.dot(SigmaReduced3.dot(VTReduced3))
print("Matriz A transformada:")
print(AReduced3)

T3 = UReduced3.dot(SigmaReduced3)
print("Matriz T (3 Dimensiones):")
print(T3)
```

```python
from mpl_toolkits.mplot3d import Axes3D

fig = plt.figure()
ax3d = fig.add_subplot(projection="3d")
arrayx = np.array(T3[:,0])
arrayy = np.array(T3[:,1])
arrayz = np.array(T3[:,2])
ax3d.scatter(arrayx, arrayy, arrayz)
for x, y, z, label in zip(arrayx, arrayy, arrayz, estudiantes):
    ax3d.text(x,y,z,label)
plt.title("Datos en 3D (SVD)")
plt.show()
```

```python
plt.figure(figsize=(10,7))

#para ñas dos dimensiones 
Error2 = A.values - AReduced2
dfError2 = pd.DataFrame(Error2, columns=A.columns)
dfError2.boxplot()
print("Errores promedio por columna (2D)")
print(Error2.mean(axis=0))

#para las tres dimendiones 
Error3 = A.values - AReduced3
print(Error3)
plt.figure(figsize=(10,7))
dfError3 = pd.DataFrame(Error3, columns=A.columns)
dfError3.boxplot()
print("Errores promedio por columna (3D)")
print(Error3.mean(axis=0))
```
